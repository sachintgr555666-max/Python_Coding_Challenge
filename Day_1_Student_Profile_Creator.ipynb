# Day 1 - Student Profile Creator


```python
# Day 1 - Student Profile Creator

name = input("Enter student name: ")

age = int(input("Enter student age: "))

subjects = input("Enter favourite subjects separated by commas: ").split(",")

subjects = [subject.strip() for subject in subjects]

print("\n--- Student Information ---")
print(f"Name: {name}")
print(f"Age: {age}")
print("Favourite Subjects:", ", ".join(subjects))

new_subject = input("\nEnter a new subject to add: ").strip()

subjects.append(new_subject)

print("\nUpdated List of Subjects:")
print(", ".join(subjects))
```
