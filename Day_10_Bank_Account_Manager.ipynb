# Day 10 Bank Account Manager


```python
# Day 10 - Bank Account Manager

transaction_count = 0


class BankAccount:
    def __init__(self, account_holder, balance=0):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        global transaction_count
        self.balance += amount
        transaction_count += 1
        print(f"Deposited: {amount:.2f}")

    def withdraw(self, amount):
        global transaction_count

        if amount > self.balance:
            print("Withdrawal denied. Insufficient balance.")
            return

        self.balance -= amount
        transaction_count += 1
        print(f"Withdrawn: {amount:.2f}")

    def display_balance(self):
        print(f"Account Holder: {self.account_holder}")
        print(f"Current Balance: {self.balance:.2f}")
```

```python
account = BankAccount("Arun", 5000)
```

```python
print("--- Initial Account ---")
account.display_balance()

print("\nUsing positional arguments:")
account.deposit(1500)

print("\nUsing keyword arguments:")
account.withdraw(amount=1000)

print("\nFinal Account Details:")
account.display_balance()

print("\nTotal Transactions:", transaction_count)
```
