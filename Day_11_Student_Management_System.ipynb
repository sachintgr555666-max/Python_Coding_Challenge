# Day 11 Student Management System


```python
# Day 11 - Student Management System

students = [
    {"name": "Arun", "age": 20, "grade": 85},
    {"name": "Bala", "age": 21, "grade": 72},
    {"name": "Cathy", "age": 20, "grade": 91}
]
```

```python
available_classes = frozenset(["Python", "SQL", "Power BI", "Excel"])
```

```python
def display_students():
    print("\n--- Student List ---")
    if not students:
        print("No students available.")
        return

    for student in students:
        print(
            f"Name: {student['name']}, "
            f"Age: {student['age']}, "
            f"Grade: {student['grade']}"
        )
```

```python
def add_student():
    try:
        name = input("Enter student name: ").strip()
        age = int(input("Enter student age: "))
        grade = float(input("Enter student grade: "))

        if not name:
            raise ValueError("Name cannot be empty.")

        if age < 0 or not 0 <= grade <= 100:
            raise ValueError("Invalid age or grade.")

        students.append({
            "name": name,
            "age": age,
            "grade": grade
        })

        print("Student added successfully.")

    except ValueError as error:
        print("Invalid input:", error)
```

```python
def delete_student():
    name = input("Enter student name to delete: ").strip()

    for student in students:
        if student["name"].lower() == name.lower():
            students.remove(student)
            print("Student deleted successfully.")
            return

    print("Student not found.")
```

```python
def update_student():
    name = input("Enter student name to update: ").strip()

    for student in students:
        if student["name"].lower() == name.lower():
            try:
                new_age = int(input("Enter new age: "))
                new_grade = float(input("Enter new grade: "))

                if new_age < 0 or not 0 <= new_grade <= 100:
                    raise ValueError("Invalid age or grade.")

                student["age"] = new_age
                student["grade"] = new_grade

                print("Student updated successfully.")

            except ValueError as error:
                print("Invalid input:", error)

            return

    print("Student not found.")
```

```python
def search_student():
    name = input("Enter student name to search: ").strip()

    for student in students:
        if student["name"].lower() == name.lower():
            print("Student found:")
            print(student)
            return

    print("Student not found.")
```

```python
def categorize_performance():
    print("\n--- Performance Report ---")

    for student in students:
        grade = student["grade"]

        if grade >= 90:
            category = "Excellent"
        elif grade >= 75:
            category = "Good"
        elif grade >= 50:
            category = "Average"
        else:
            category = "Needs Improvement"

        print(f"{student['name']}: {grade} - {category}")
```

```python
def sort_students():
    students.sort(key=lambda student: student["grade"], reverse=True)

    print("\nStudents sorted by grade:")
    for student in students:
        print(student)
```

```python
def filter_students():
    try:
        minimum_grade = float(input("Enter minimum grade for filtering: "))

        filtered_students = [
            student for student in students
            if student["grade"] >= minimum_grade
        ]

        print("\nFiltered Students:")
        for student in filtered_students:
            print(student)

    except ValueError:
        print("Please enter a valid number.")
```

```python
def main():
    print("=== Student Management System ===")
    print("Available Classes:", available_classes)

    while True:
        print("\n1. Display students")
        print("2. Add student")
        print("3. Delete student")
        print("4. Update student")
        print("5. Search student")
        print("6. Performance report")
        print("7. Sort students")
        print("8. Filter students")
        print("9. Exit")

        try:
            choice = int(input("Enter your choice: "))

            if choice == 1:
                display_students()
            elif choice == 2:
                add_student()
            elif choice == 3:
                delete_student()
            elif choice == 4:
                update_student()
            elif choice == 5:
                search_student()
            elif choice == 6:
                categorize_performance()
            elif choice == 7:
                sort_students()
            elif choice == 8:
                filter_students()
            elif choice == 9:
                print("Student Management System closed.")
                break
            else:
                print("Invalid choice. Please select 1 to 9.")

        except ValueError:
            print("Invalid input. Please enter a number.")
```

```python
main()
```
