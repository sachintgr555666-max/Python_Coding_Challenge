```python
# Day 3 - Retail Inventory Manager

categories = ("Electronics", "Stationery", "Clothing")

products = ["Laptop", "Notebook", "T-Shirt"]

print("Product Categories:", categories)
print("Initial Products:", products)

new_product = input("\nEnter a new product: ").strip()
products.append(new_product)

print("Products after adding:", products)

print("\nCurrent product list:")
for index, product in enumerate(products):
    print(index, "-", product)

index = int(input("Enter the index of the product to remove: "))

if 0 <= index < len(products):
    removed_product = products.pop(index)
    print(f"{removed_product} removed successfully.")
else:
    print("Invalid index. No product was removed.")

print("\nFinal List of Products:")
print(products)

print("\nProduct Categories:")
print(categories)
```
