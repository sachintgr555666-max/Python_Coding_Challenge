# Day 5 Student Grades Hobbies Tracker


```python
# Day 5 - Student Grades & Hobbies Tracker

grades = {
    "Arun": 85,
    "Bala": 78,
    "Cathy": 92
}

print("--- Initial Grades ---")
print(grades)
```

```python
for i in range(3):
    name = input(f"\nEnter student {i + 1} name: ").strip()
    grade = float(input(f"Enter grade for {name}: "))
    grades[name] = grade

print("\nUpdated Student Grades:")
for name, grade in grades.items():
    print(f"{name}: {grade}")
```

```python
hobbies = set()

print("\nEnter at least 3 hobbies:")
for i in range(3):
    hobby = input(f"Enter hobby {i + 1}: ").strip()
    hobbies.add(hobby)

hobbies_frozenset = frozenset(hobbies)

print("\nHobby Set:")
print(hobbies)

print("\nHobbies as Frozenset:")
print(hobbies_frozenset)
```
