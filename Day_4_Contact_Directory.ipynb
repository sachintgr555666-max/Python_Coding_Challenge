# Day 4 Contact Directory


```python
# Day 4 - Contact Directory

contacts = [
    ["Arun", "9876543210", "arun@example.com"],
    ["Bala", "9876543211", "bala@example.com"],
    ["Cathy", "9876543212", "cathy@example.com"]
]
```

```python
print("--- Initial Contacts ---")
for contact in contacts:
    print(contact)
```

```python
print("\nAdd a new contact")
name = input("Enter name: ").strip()
phone = input("Enter phone number: ").strip()
email = input("Enter email address: ").strip()

contacts.append([name, phone, email])

print("\nContacts after adding new contact:")
for contact in contacts:
    print(contact)
```

```python
removed_contact = contacts.pop()
print("\nRemoved last item:", removed_contact)
```

```python
print("\nEnter another contact")
name = input("Enter name: ").strip()
phone = input("Enter phone number: ").strip()
email = input("Enter email address: ").strip()

contacts.append([name, phone, email])
```

```python
contacts.sort(key=lambda contact: contact[0])

print("\nFinal Contact Directory (sorted by name):")
for contact in contacts:
    print(contact)
```
